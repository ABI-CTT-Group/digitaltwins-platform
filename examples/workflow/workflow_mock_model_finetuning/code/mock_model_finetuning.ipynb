## Write the output

Copy the input dataset to `outputs/datasets/finetuned_models/` with a note that it is a mock output, so Submit uploads a dataset.

```python
sources = sorted(p.parent for p in INPUTS_DIR.glob("*/dataset_description.xlsx"))
assert sources, f"no input dataset found under {INPUTS_DIR}/"
target = OUTPUTS_DIR / OUTPUT_NAME
if target.exists():
    shutil.rmtree(target)
shutil.copytree(sources[0], target, ignore=shutil.ignore_patterns(".ipynb_checkpoints"))
(target / "docs").mkdir(exist_ok=True)
(target / "docs" / "mock_output.txt").write_text(
    f"{OUTPUT_NAME} from a mock notebook: a copy of the input dataset; nothing was computed.\n"
)
print("wrote", target)
```
